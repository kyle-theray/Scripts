```python
import arcpy
import os
from datetime import datetime

# --------------------------------------------------------------------------
# USER INPUTS
# --------------------------------------------------------------------------
input_raster = r"C:\Exclusion_Analysis_Phase3\Input_data\conus_composite_solar_exclusion_10m_5070.tif"  # pre-processed inclusion raster
output_gdb = r"C:\Exclusion_Analysis_Phase3\output.gdb"
final_output_fc = os.path.join(output_gdb, "CONUS_Revised_ROW_10m_Clipped")

# --------------------------------------------------------------------------
# ENVIRONMENT
# --------------------------------------------------------------------------
arcpy.env.overwriteOutput = True

# --------------------------------------------------------------------------
# LOG FUNCTION
# --------------------------------------------------------------------------
log_file = os.path.join(output_gdb, f"raster_to_poly_log_{datetime.now().strftime('%Y%m%d_%H%M%S')}.txt")
def log(msg):
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{timestamp}] {msg}")
    with open(log_file, "a", encoding="utf-8") as f:
        f.write(f"[{timestamp}] {msg}\n")

log("=== SCRIPT START ===")
log(f"Input Raster: {input_raster}")
log(f"Output GDB: {output_gdb}")

# --------------------------------------------------------------------------
# RASTER TO POLYGON
# --------------------------------------------------------------------------
log("Converting raster to polygons...")
start_time = datetime.now()

temp_poly = os.path.join(output_gdb, "temp_raster_poly")
arcpy.conversion.RasterToPolygon(input_raster, temp_poly, "SIMPLIFY", "Value")

elapsed = datetime.now() - start_time
log(f"Raster to polygon completed in {elapsed}")

# --------------------------------------------------------------------------
# FINAL DISSOLVE
# --------------------------------------------------------------------------
log("Performing final dissolve to merge polygons...")
start_time = datetime.now()

arcpy.analysis.PairwiseDissolve(temp_poly, final_output_fc)

elapsed = datetime.now() - start_time
log(f"Final dissolve completed in {elapsed}")
log(f"Final output: {final_output_fc}")

# --------------------------------------------------------------------------
# CLEANUP
# --------------------------------------------------------------------------
arcpy.management.Delete(temp_poly)

log("=== SCRIPT END ===")
```
